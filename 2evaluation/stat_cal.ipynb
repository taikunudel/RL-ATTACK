```python
import numpy as np
import json
def average_mean_and_se_folds(data, n_folds=10, var_name=''):
    """
    Splits the data into n_folds, calculates the mean and sample standard error (SE)
    for each fold, and returns the average mean and the average SE across the folds.
    
    Parameters:
        data (list or np.ndarray): The list or array of numbers.
        n_folds (int): Number of folds to split the data into (default is 10).

    Returns:
        tuple: (average mean, average standard error) across folds.
    """
    # Convert data to a numpy array if it is not already
    data = np.array(data)
    print(len(data))
    # Split the data into nearly equal folds
    folds = np.array_split(data, n_folds)
    
    means = []
    se_list = []
    for fold in folds:
        # Compute the mean of the fold
        mean_fold = np.mean(fold)
        means.append(mean_fold)
        
        # If a fold has only one element, the standard error is set to 0
        if len(fold) <= 1:
            se = 0.0
        else:
            # Calculate the sample standard deviation (ddof=1 for sample SD)
            s = np.std(fold, ddof=1)
            se = s / np.sqrt(len(fold))
        se_list.append(se)
    
    avg_mean = np.mean(means)
    avg_se = np.mean(se_list)
    
    print("-------", var_name, "-------")
    print("Mean: {:.4f}".format(avg_mean))
    print("Standard Error: {:.6f}".format(avg_se))    
    return avg_mean, avg_se

# file_name = '/usa/taikun/rl-attack/2evaluation/snli/premise/eva_0923_105240_premise_BERTFineTuned_BERT_SNLI_snli_val_0.7_8_60_0.json'
file_name = '/usa/taikun/rl-attack/2evaluation/snli/premise/eva_0923_150836_premise_BERTMaskedLM_BERT_SNLI_snli_val_0.7_8_60_0.json'

with open(file_name, 'r') as f:
    data = json.load(f)

# src_docs = [d[0] for d in data['sourceDocs']]
# adv_docs = [d[0] for d in data['advDocs']]
success_flags = data["attackSuccess Flags"]                          # Binary success list

doc_lens = data["docLen"]                                            # Total length of each doc
tgt_tokens = [len(tok) for tok in data["targetTokens"]]              # Perturbed token count per doc
perturb_ratios = [p / l if l > 0 else 0 for p, l in zip(tgt_tokens, doc_lens)]

average_mean_and_se_folds(data = success_flags, var_name='mean')
average_mean_and_se_folds(data = perturb_ratios, var_name='pertb')

# similarities = 
average_mean_and_se_folds(data = similarities, var_name='USE')


```
