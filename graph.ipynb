```python
import matplotlib.pyplot as plt
import pandas as pd

# Data for SNLI Premise
snli_premise_data = {
    'max_allowed_queries': [20, 40, 60, 80, 100, 120],
    'accAftAtk_premise': [0.187, 0.146, 0.121, 0.106, 0.101, 0.091],
    'actual_queries_premise': [24.394, 30.163, 33.498, 38.146, 43.455, 45.826]
}

# Data for SNLI Hypothesis
snli_hypothesis_data = {
    'max_allowed_queries': [20, 40, 60, 80, 100, 120],
    'accAftAtk_hypothesis': [0.0644, 0.0203, 0.019, 0.013, 0.01, 0.006],
    'actual_queries_hypothesis': [11.62042175, 13.10885341, 14.71, 15.981, 17.54, 18.157]
}

# Create two subplots for Premise and Hypothesis, horizontally aligned, with reduced height
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5), sharex=True)

# Plot for SNLI Premise
ax1.set_title('SNLI Premise')
ax1.set_xlabel('Maximum Allowed Queries')
ax1.set_ylabel('Accuracy After Attack', color='tab:blue')
ax1.plot(snli_premise_data['max_allowed_queries'], snli_premise_data['accAftAtk_premise'], label='Accuracy', color='tab:blue', marker='o')
ax1.tick_params(axis='y', labelcolor='tab:blue')

# Second y-axis for actual queries
ax1a = ax1.twinx()
ax1a.set_ylabel('Actual Queries', color='tab:red')
ax1a.plot(snli_premise_data['max_allowed_queries'], snli_premise_data['actual_queries_premise'], label='Actual Queries', color='tab:red', linestyle='--', marker='x')
ax1a.tick_params(axis='y', labelcolor='tab:red')

# Plot for SNLI Hypothesis
ax2.set_title('SNLI Hypothesis')
ax2.set_xlabel('Maximum Allowed Queries')
ax2.set_ylabel('Accuracy After Attack', color='tab:blue')
ax2.plot(snli_hypothesis_data['max_allowed_queries'], snli_hypothesis_data['accAftAtk_hypothesis'], label='Accuracy', color='tab:blue', marker='o')
ax2.tick_params(axis='y', labelcolor='tab:blue')

# Second y-axis for actual queries
ax2a = ax2.twinx()
ax2a.set_ylabel('Actual Queries', color='tab:red')
ax2a.plot(snli_hypothesis_data['max_allowed_queries'], snli_hypothesis_data['actual_queries_hypothesis'], label='Actual Queries', color='tab:red', linestyle='--', marker='x')
ax2a.tick_params(axis='y', labelcolor='tab:red')

# Adjust layout and show
plt.tight_layout()
plt.show()
```
